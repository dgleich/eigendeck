# Interactive Plotly test

This notebook has a **baked** Plotly output (loads plotly.js from the CDN), so it
renders without a running kernel. In Eigendeck it must mount in a sandboxed
iframe and stay fully interactive: drag to zoom, hover for tooltips, click the
legend to toggle traces. It must NOT be able to reach the app / filesystem.


```python
import plotly.graph_objects as go
fig = go.Figure()
fig.add_scatter(x=list(range(1,11)), y=[i*i for i in range(1,11)], name='x squared')
fig.add_scatter(x=list(range(1,11)), y=[2**i for i in range(1,11)], name='2 to the x')
fig.update_layout(title='Interactive Plotly')
fig.show()
```
